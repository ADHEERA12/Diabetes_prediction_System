```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn import svm
from sklearn.metrics import accuracy_score
import pickle
```

Data Collectiob ad Analysis




```python
diabetes_data=pd.read_csv("/content/diabetes (3).csv")
diabetes_data.head()
#1-yes diabetic ,0-no debetic
```

```python
#no of rows and columns
diabetes_data.shape
```

```python
#statistical measures of the data
diabetes_data.describe()#gives the count ,mean ,sd,min,max,etc..
```

```python
#finding the toal number of each type of outcome
diabetes_data['Outcome'].value_counts()
```

```python
diabetes_data.groupby('Outcome').mean()
```

```python
#seperating the data and labels
X=diabetes_data.drop(columns="Outcome",axis=1);#drop(colums,axis=0 mean we r deleting row,1 mean we r deleting column)
Y=diabetes_data['Outcome'];
print(X);
```

```python
print(Y)
```

```python
#Standardize the data
scaler=StandardScaler()
scaler.fit(X)
```

#Standardized data contains values only within 0 and 1

```python
standardized_data=scaler.transform(X);
```

```python
print(standardized_data)
```

```python
X=standardized_data;
```

```python
#train -test split(4 variables)
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,stratify =Y,random_state=2)
#test_size=20%percent for test
#random_state=means:
# Use seed value 2 for randomization.
# Every time you run the code, you get the same train-test split.


#stratify -ensures that both the test and the train data have better data distribution ,x train-5 yes 5 no y train -2 yes 2 no
```

```python
print(X_train.shape,X_test.shape)
```

TRAINING THE MODEL

```python

#Support vector classifier
classifier=svm.SVC(kernel='linear')


```

```python
#training the svm
classifier.fit(X_train,Y_train)
```

```python
#Model Evaluation
#1.Accuracy Score
#using svm to predict the trained data
X_train_prediction=classifier.predict(X_train)
training_data_Accuracy=accuracy_score(X_train_prediction,Y_train)
print("Accuracy score of the training data :",training_data_Accuracy)

```

Accuracyy sicore of the test data


```python
X_test_prediction=classifier.predict(X_test)
test_data_Accuracy=accuracy_score(X_test_prediction,Y_test)
print("Accuracy score of the test data :",test_data_Accuracy)
```
